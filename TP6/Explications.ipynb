On partage les données en 3 parties : 
1) les données d'apprentissage ~ environ 60-70% de l'ensemble des données
2) les données de validation ~ 10-20% de l'ensemble des données : 
    elles servent à déterminer les hyperparamètres (nb de neurones, nb de couches, pas d'apprentissage, paramètre de régularisation)
3) les données test ~ 10-20% de l'ensemble des données :
    elles servent à tester le modèle final avec les meilleurs hyperparamètres

Les données test doivent être utilisées le moins possible, idéalement une seule fois
